### Ways of Handling Missing Values

Missing values will be handled differently depending on the variable.

- Categorical variables will be filled only when appropriate.
- Records missing essential analytical information will be removed.
- Numeric variables will be preserved unless transformation was required.

##### Handling Missing Values in "RemoteWork" Column

```python
df_clean["RemoteWork"] = (
    df_clean["RemoteWork"]
    .fillna("Unknown")
)
```

RemoteWork column includes 16.2% of missing vaues. As this column consisats of categorical values (string data type), the missing values in the column are marked as "Unknown". Otherwise coming up with absolutely new data might change the overall outcome of the dataset.

```python
df_clean["RemoteWork"].value_counts(dropna=False)
```

##### Handling Missing Values in "JobSat" Column

JobSat column represents numeric values (float64 data type). Sinse job satisfaction is one of the central outcomes of this analysis, records without a `JobSat` value cannot be used in satisfaction-related comparisons.

These records are removed from the analytical dataset. However, because more than half of the original responses are excluded (36311 values or 55.1%), the remaining sample (29126 rows) may not fully represent all survey respondents. This limitation will be considered when interpreting the results.

```python
df_clean = df_clean.dropna(
    subset=["JobSat"]
)
```

```python
print(df_clean.shape)
```

```python
rows_before = len(df)
rows_after = len(df_clean)
rows_removed = rows_before - rows_after
retained_percentage = rows_after / rows_before * 100

print(f"Rows before cleaning: {rows_before:,}")
print(f"Rows after cleaning: {rows_after:,}")
print(f"Rows removed: {rows_removed:,}")
print(f"Data retained: {retained_percentage:.1f}%")
```

After removing records with missing job satisfaction values, the analytical dataset contains 29,126 rows and 114 columns.
All remaining records contain a valid `JobSat` response.

##### Handling Missing Values in "Employment" Column

No missing values were identified in the Employment variable.

##### Transformation in "YearsCodePro" Column

The `YearsCodePro` variable contains text values,as well as two special categorical responses. That must be converted into a numeric format before statistical analysis.

```python
df_clean["YearsCodePro"].value_counts(dropna=False).head(20)
```

```python
sorted(df_clean["YearsCodePro"].dropna().unique())
```

```python
experience_mapping = {
    "Less than 1 year": 0.5,
    "More than 50 years": 51
}

df_clean["YearsCodePro"] = (
    df_clean["YearsCodePro"]
        .replace(experience_mapping)
)
```

```python
df_clean["YearsCodePro"] = pd.to_numeric(
    df_clean["YearsCodePro"],
    errors="coerce"
)
```

```python
df_clean["YearsCodePro"].describe()
```

The `YearsCodePro` variable originally contained both numeric and text values.

Special responses such as **"Less than 1 year"** and **"More than 50 years"** were converted into numeric values to enable statistical analysis and visualization.

"Less than 1 year" is represented as 0.5,  "More than 50 years"` is represented as 51. These values are proxies used to preserve the ordering of experience levels and should not be interpreted as exact measurements.

The variable is now suitable for quantitative analysis.

### Analysis 2 — Job Satisfaction

#### Research Question 2

How satisfied are developers with their current jobs?

```python
job_sat = (
    df_clean["JobSat"]
    .value_counts()
    .sort_index()
)

job_sat_percentage = (
    df_clean["JobSat"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(1)
)

job_sat_summary = pd.DataFrame({
    "Respondents": job_sat,
    "Percentage": job_sat_percentage
})

job_sat_summary
```

#### Conclusions

Job satisfaction varies across respondents. The chart identifies whether developers tend to report positive, neutral, or negative satisfaction levels. Most respondents classified their job satisfaction as 7.0 (6379 answers or 21.9%) and 8.0 (7509 answers or 25.8%). This assessment can be characterized above average which might be considered as a good indicator. It should also be noted that 3626 respondents(12.4%) evaluated job satisfaction with 9.0 grade and more than 2251 respondents (7.7%) with 10.0 grade.

#### Research Question 5

Which programming languages are most commonly used by developers?

To get the actual data for analysis, the column "LanguageHaveWorkedWith" must be splited by symbol ";" as it contains not only one programming language in the cell but the list of languages. If the separation is not performed, the result will be a ranking of language combinations, rather than of the languages ​​themselves.

```python
languages = (
    df_clean["LanguageHaveWorkedWith"]
        .dropna()
        .str.split(";")
        .explode()
)
```

```python
languages.head(20)
```

```python
language_counts = (
    languages
        .value_counts()
)
language_counts.head(20)
```

```python
top_languages = (
    language_counts
        .head(15)
)
```

```python
plt.figure(figsize=(11,7))

sns.barplot(
    x=top_languages.values,
    y=top_languages.index
)

plt.title("Top 15 Programming Languages Used by Developers")

plt.xlabel("Number of Respondents")

plt.ylabel("Programming Language")

plt.tight_layout()

plt.show()
```

#### Conclusion

JavaScript, Python, SQL, and HTML/CSS are among the most frequently used programming languages reported by survey respondents.

#### Research Question 6

Which database technologies are most commonly used by developers?

DatabaseHaveWorkedWith column represents the same type of data as ProgrammingLanguagesHaveWorkedWith - the list of values in one cell. It works like that because survey participants were able to respond multiple answers on the question.
To perform the appropriate analysis the column must be splited by symbol ";" as well.

```python
databases = (
    df_clean["DatabaseHaveWorkedWith"]
        .dropna()
        .str.split(";")
        .explode()
)
```

```python
databases.head(20)
```

```python
database_counts = (
    databases
        .value_counts()
)
```

```python
database_counts.head(20)
```

```python
top_databases = (
    database_counts
        .head(15)
)
```

```python
plt.figure(figsize=(11,7))

sns.barplot(
    x=top_databases.values,
    y=top_databases.index
)

plt.title("Top 15 Database Technologies Used by Developers")

plt.xlabel("Number of Respondents")

plt.ylabel("Database Technology")

plt.tight_layout()

plt.show()
```

```python
database_summary = pd.DataFrame({
    "Respondents": database_counts
})

database_summary["Percentage"] = (
    database_summary["Respondents"]
    / len(df_clean)
    * 100
).round(1)

database_summary.head(15)
```

#### Conclusion

Relational database management systems remain the most widely used technologies among survey respondents, with PostgreSQL (47.2% of respondents), MySQL (35.1% of respondents), SQLite (29.2% of respondents), and Microsoft SQL Server (25.9% of respondents) appearing frequently in developers' technology stacks.

At the same time, modern NoSQL solutions such as MongoDB are also widely adopted, demonstrating the increasing diversity of database technologies used in software development.

Because respondents could select multiple databases, the reported percentages represent the share of developers who use each technology rather than mutually exclusive categories.

#### Research Question 8

Is there a relationship between professional coding experience and job satisfaction?

```python
experience_jobsat = (
    df_clean[
        ["YearsCodePro", "JobSat"]
    ]
    .dropna()
    .copy()
)
```

```python
experience_jobsat.head()
```

```python
jobsat_by_experience = (
    experience_jobsat
        .groupby("YearsCodePro")["JobSat"]
        .mean()
        .reset_index()
)
```

```python
jobsat_by_experience.head()
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=jobsat_by_experience,
    x="YearsCodePro",
    y="JobSat",
    marker="o"
)

plt.title("Average Job Satisfaction by Professional Coding Experience")

plt.xlabel("Years of Professional Coding")

plt.ylabel("Average Job Satisfaction")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

Since the line plot reflects the average satisfaction trend in relation to coding experience—and significant fluctuations in the line are observed after 35 years—it is necessary to assess the distribution of all observations, specifically their frequency. This makes it possible to account for the impact of sample size for each category YearsCodePro and to compare the volume of response variations included in the analysis. Building a scatter plot is the best way to check it out.

```python
plt.figure(figsize=(12,6))

sns.scatterplot(
    data=experience_jobsat,
    x="YearsCodePro",
    y="JobSat",
    alpha=0.15
)

plt.title("Distribution of Respondent Answers by Professional Coding Experience vs Job Satisfaction")
plt.xlabel("Years of Professional Coding")
plt.ylabel("Job Satisfaction")

plt.tight_layout()
plt.show()
```

On the scatter chart we can clearly see that the amount of respondents' answers decrease with the growth of coding experience. It means that the amount of observations with high code experience is much less then the observations with 0-30 years of coding experience which could likely lead to significant fluctuations in the responds.

```python
jobsat_by_experience.describe()
```

```python
experience_summary = (
    experience_jobsat
        .groupby("YearsCodePro")
        .agg(
            Average_JobSat=("JobSat", "mean"),
            Respondents=("JobSat", "count")
        )
)

experience_summary.tail(25)
```

#### Conclusion

Average job satisfaction remains relatively stable across most experience levels, with only moderate variation over respondents' professional careers.

Some fluctuations appear among respondents with very high levels of experience. However, these values should be interpreted cautiously because they are based on a relatively small number of observations. For example, there were only 44 observations for 37 years of coding experience, 13 odservations for 44 years against 212 obserations for 27 years and 100 for 32 years of coding experience.

Overall, the results suggest that professional experience alone is not a strong determinant of job satisfaction, indicating that other factors may have a greater influence on how satisfied developers are with their work.

# Summary and Key Findings

This project analyzed responses from the Stack Overflow Developer Survey to better understand developers' backgrounds, technology preferences, and factors related to job satisfaction.

The dataset was cleaned, transformed, and explored using Python data analysis libraries. Exploratory analysis focused on professional experience, education, work arrangements, programming languages, database technologies, and the relationship between experience and job satisfaction.

### Key Findings

- Most respondents reported fewer than 10 years of professional coding experience.
- Job satisfaction ratings were generally concentrated in the upper half of the scale.
- Hybrid and remote work arrangements were more common than fully in-person work.
- Bachelor's and Master's degrees represented the largest educational groups.
- JavaScript, SQL, HTML/CSS, and Python were among the most widely used programming languages.
- PostgreSQL, MySQL, SQLite, and Microsoft SQL Server were among the most commonly used database technologies.
- Professional coding experience showed only a modest relationship with average job satisfaction, suggesting that additional factors are likely to influence developers' satisfaction at work.

### Limitations

Several limitations should be considered when interpreting the results:

- Records with missing job satisfaction values were excluded from analyses involving `JobSat`.
- The survey is based on self-reported responses and may contain reporting bias.
- Some survey questions allowed multiple selections, meaning percentages may exceed 100% (for example, database technologies, programming languages).
- The findings describe this survey sample and should not be interpreted as representing all software developers worldwide.
